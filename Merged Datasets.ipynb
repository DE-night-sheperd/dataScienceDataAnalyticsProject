# 3. HICP Food, Alcohol & Tobacco


## Exploratory Data Analysis (EDA)

HICP Food Alcohol Tobacco

This notebook explores the dataset by performing data inspection, cleaning, feature engineering, visualization, and analysis to uncover meaningful insights and gain a deeper understanding of the data.

Dataset Description (Write-up)

The dataset used in this analysis is the Harmonised Index of Consumer Prices (HICP) dataset, specifically focusing on food, alcohol, and tobacco price indices across multiple countries. This dataset was obtained from an open data portal (Eurostat), which provides reliable and standardized economic indicators across countries.

The dataset contains monthly index values that measure changes in the prices of selected consumer goods over time. The key variables in the dataset include:

freq: Frequency of data (monthly)
geo: Country identifier
coicop18: Product category (e.g., food, alcohol, tobacco)
unit: Measurement unit (index values)
Time columns (e.g., 2025-01 to 2026-01): Monthly price index values

This dataset is suitable for the problem because it allows analysis of inflation trends and price changes over time, enabling comparisons across countries and identifying patterns in consumer price movements.

#Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

The selected libraries support efficient data manipulation and visualization. pandas enables structured data handling, while matplotlib facilitates graphical exploration, which is essential for identifying patterns, trends, and anomalies in time-series economic data.

#Load Dataset

```python
df = pd.read_csv("HICP-food-alcohol-tobacco.csv")
df.head()
```

The dataset loads successfully and reveals a wide-format structure, where each column represents a different time period. This structure is not optimal for time-series analysis, indicating the need for transformation into a long (tidy) format.

#Inspect Dataset

```python
df.info()
df.describe()
```

The dataset consists of both categorical variables (e.g., country and product category) and numerical variables (monthly index values). However, some columns contain non-numeric entries, suggesting potential data quality issues that must be addressed before analysis. The presence of multiple time-based columns confirms that the dataset captures temporal price dynamics.

#Check Missing Values

```python
df.replace(":", np.nan, inplace=True)
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

The presence of the symbol ":" indicates systematically missing observations rather than random absence of data. If left untreated, these values would interfere with numerical computations and bias summary statistics, making it essential to standardize them as proper missing values.

Converting placeholder symbols to NaN ensures compatibility with Python’s data analysis functions. This step is critical because most statistical and visualization tools in Python automatically handle NaN values appropriately, preventing distortions in the analysis.

#Convert Numeric Columns

```python
for col in df.columns[4:]:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

Converting the date column into a proper datetime format allows for chronological ordering and time-based operations. This is essential for identifying trends, seasonality, and temporal patterns in price indices.

#Reshape Data (Wide to Long)

```python
df_long = df.melt(id_vars=['freq','geo','coicop18','unit'],
                  var_name='Date',
                  value_name='Value')

df_long.head()
```

Restructuring the dataset into a long format significantly improves its usability for analysis. This transformation aligns with the principles of tidy data, where each row represents a single observation. It also enables more flexible and efficient time-series visualizations and grouping operations.

#Convert Date Column

```python
df_long['Date'] = pd.to_datetime(df_long['Date'])
```

Converting the date column into a proper datetime format allows for chronological ordering and time-based operations. This is essential for identifying trends, seasonality, and temporal patterns in price indices.

#Line Plot (Trend Over Time)

```python
plt.figure()

for country in df_long['geo'].unique()[:5]:
    plt.plot(
        df_long[df_long['geo'] == country]['Date'],
        df_long[df_long['geo'] == country]['Value']
    )

plt.xlabel("Date")
plt.ylabel("Index Value")
plt.title("Price Trends Over Time")

plt.show()
```

The line plot reveals clear temporal trends in price indices across countries. Most countries exhibit a gradual upward trajectory, which is consistent with inflationary behavior over time. Differences in slope suggest varying inflation rates between countries, while any sharp fluctuations may indicate economic shocks or short-term price volatility.

```python
#Boxplot
```

```python
plt.figure()
df_long.boxplot(column='Value')
plt.title("Distribution of Price Index Values")
plt.show()
```

The boxplot highlights the distribution and dispersion of price index values. The presence of outliers suggests periods of unusually high or low price changes, potentially linked to external economic factors such as supply disruptions or policy changes. The spread of the data also indicates variability in inflation levels across countries and time periods.

#Histogram

```python
plt.figure()
df_long['Value'].hist()
plt.title("Distribution of Values")
plt.xlabel("Index Value")
plt.ylabel("Frequency")
plt.show()
```

The histogram provides insight into the overall distribution of price index values. A skewed distribution may indicate that most observations cluster within a specific range, while extreme values occur less frequently. This helps in understanding whether price changes are consistent or influenced by occasional spikes.

#Correlation Matrix

```python
df_corr = df.iloc[:, 4:].corr()

plt.figure()
plt.imshow(df_corr)
plt.colorbar()
plt.title("Correlation Matrix")
plt.show()
```

The correlation matrix reveals strong positive relationships between adjacent time periods, which is expected in economic time-series data. This suggests that price indices change gradually over time rather than abruptly. High correlations confirm temporal stability, while lower correlations (if any) may indicate structural changes or economic disruptions.

#Heatmap Style View

```python
plt.figure()
plt.imshow(df_corr)
plt.title("Heatmap of Correlations")
plt.colorbar()
plt.show()
```

The heatmap visually reinforces the correlation findings, with stronger correlations appearing along the diagonal. This pattern indicates that values closer in time are more closely related, which is characteristic of inflation data. The visualization also helps identify any irregularities in temporal relationships.

#Country Comparison

```python
avg_values = df_long.groupby('geo')['Value'].mean()

plt.figure()
avg_values.plot(kind='bar')
plt.title("Average Price Index by Country")
plt.show()
```

The bar chart comparing average price indices across countries highlights relative differences in inflation levels. Countries with higher average index values may be experiencing sustained price increases, while lower values indicate more stable price conditions. This comparison is useful for identifying economic disparities and regional inflation trends.


## Overall Insight:

The exploratory data analysis reveals that price indices for food, alcohol, and tobacco exhibit consistent upward trends across most countries, reflecting inflationary pressures. The strong correlations across time confirm the persistence of price movements, while variations between countries highlight differences in economic conditions. The presence of outliers suggests that certain periods experienced abnormal price changes, likely driven by external economic factors.
